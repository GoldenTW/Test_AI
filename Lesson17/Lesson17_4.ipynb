```python
import pandas as pd

df = pd.DataFrame({
    '國文': [80, 90, 70],
    '英文': [85, 88, 75],
    '數學': [90, 92, 78]
})

df.apply(lambda x: x.mean()) #每一科目的平均分數
```

```python
df.apply(lambda x: x.sum(), axis=1) #每位學生的總分
```

```python
def judge(row):
    if row['國文'] < 60:
        return '國文不及格'
    elif row['數學'] < 60:
        return '數學不及格'
    else:
        return '全部及格'

df['評語'] = df.apply(judge, axis=1)

df
```
