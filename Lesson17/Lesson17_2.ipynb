```python
import pandas as pd

world_df = pd.read_csv('world.csv')
print(world_df.columns)
print(world_df.info())
```

```python
world_df1 = world_df[['洲名','國家', '日期','總確診數','新增確診數','總人口數','總死亡數']]
world_df1
```

```python
mask_taiwan = world_df1["國家"] == "Taiwan"
taiwan_data = world_df1[mask_taiwan]
taiwan_data
```

```python
#使用DataFrame的Query方法
taiwan_date_query = world_df1.query("國家 == 'Taiwan'")
taiwan_date_query
```

```python
taiwan_max = taiwan_data["新增確診數"].max()
taiwan_max
taiwan_max_row = taiwan_data.query("新增確診數 == @taiwan_max")
taiwan_max_row
```
