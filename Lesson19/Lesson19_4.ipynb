```python
import pandas as pd

final_dateframe['trnOpDate'] = pd.to_datetime(
    final_dateframe['trnOpDate'].astype(str),
    format='%Y%m%d'
)
final_dateframe.head()
```

```python
final_dateframe.groupby('車站名稱', as_index=False).agg({
    'gateInComingCnt': 'mean',
    'gateOutGoingCnt': 'mean'
})
```

```python
# 如何groupby多欄位,依據年份gropby
final_dateframe['year'] = final_dateframe['trnOpDate'].dt.year
df1 = final_dateframe.groupby(['year','車站名稱'])[
    ['gateInComingCnt','gateOutGoingCnt']].sum()    
df1
```

```python
df2 = df1.unstack(level="year")
df2.head()
```
