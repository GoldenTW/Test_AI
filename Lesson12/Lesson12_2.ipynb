```python
import matplotlib.pyplot as plt

x = [0, 1, 2]
y = [3, 4, 5]

plt.figure()        # 單張圖
plt.plot(x, y, marker="o", linestyle="--", linewidth=2)  # 連線 + 標記點
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Line through (0,3), (1,4), (2,5)")

plt.show()
```

```python
import matplotlib.pyplot as plt

# 建立整張圖
fig = plt.figure(figsize=(12, 4))

# ---------------------------
# Ax 1: 折線圖（藍色）
# ---------------------------
ax_1 = fig.add_subplot(1, 3, 1)
x1 = [0, 1, 2, 3]
y1 = [2, 3, 5, 4]
ax_1.plot(x1, y1, marker="o", color="blue", linewidth=2)
ax_1.set_title("Line Chart (Blue)")

# ---------------------------
# Ax 2: 長條圖（紅色）
# ---------------------------
ax_2 = fig.add_subplot(1, 3, 2)
x2 = ["A", "B", "C"]
y2 = [5, 2, 7]
ax_2.bar(x2, y2, color="red")
ax_2.set_title("Bar Chart (Red)")

# ---------------------------
# Ax 3: 散佈圖（綠色）
# ---------------------------
ax_3 = fig.add_subplot(1, 3, 3)
x3 = [0.5, 1.5, 2.5, 3.5]
y3 = [3, 5, 2, 6]
ax_3.scatter(x3, y3, color="green", s=80)
ax_3.set_title("Scatter Plot (Green)")

# 顯示圖
plt.show()
```

```python
import matplotlib.pyplot as plt

# ==============================
#  第一張畫布：折線圖（藍色、方形 marker）
# ==============================

# 1. 製作 figure
fig1 = plt.figure()

# 2. 在 figure 上製作 axes
ax_1 = fig1.add_subplot(1, 1, 1)

# 3. 將圖的資料放入 axes 中
X1 = [0, 1, 2]
Y1 = [3, 4, 5]

ax_1.plot(
    X1, Y1,
    color='blue',
    marker='s',          # 方形 marker
    linestyle='solid',   # 實線
    linewidth=2,
    markersize=8
)

ax_1.set_title("Figure 1 - Blue Solid Line")

# 4. 顯示第一張圖
plt.show()


# ==============================
#  第二張畫布：虛線 + 圓點 marker（紅色）
# ==============================

# 1. 重新製作另一個 figure
fig2 = plt.figure()

# 2. 在第二個 figure 上製作 axes
ax_2 = fig2.add_subplot(1, 1, 1)

# 3. 放入第二組資料
X2 = [0, 1, 2]
Y2 = [5, 4, 6]

ax_2.plot(
    X2, Y2,
    color='red',
    marker='o',        # 圓形 marker
    linestyle='--',    # 虛線
    linewidth=3,
    markersize=10
)

ax_2.set_title("Figure 2 - Red Dashed Line")

# 4. 顯示第二張圖
plt.show()
```
