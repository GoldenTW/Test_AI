```python
import requests
from requests import HTTPError,JSONDecodeError
from pprint import pprint

def get_aqi() -> list[dict]:
    try: 
        url = "https://data.moenv.gov.tw/api/v2/aqx_p_488?api_key=b7df779e-71a6-4148-8379-5afbd441d803&limit=1000&sort=datacreationdate%20desc&format=JSON"
        response = requests.get(url,verify=False)
        response.raise_for_status()
        data_list:list = response.json()['records'] #data_list是每個站點的資料
        return data_list
    except HTTPError as e:
        raise Exception(f"HTTPError:{e}")
    except JSONDecodeError:
        raise Exception("內容不是JSON的格式")
    except Exception:
        raise Exception("不知明錯誤")

def filter_data(site_list:list[dict])->list[dict]:
    '''過濾資料,只要sitename,country,aqi,status'''
    filter_data:list = []
    for site in site_list:
        new_site: dict = dict()
        new_site["站點名稱"] = site.get("sitename", "")
        new_site["城市"] = site.get("county", "")
        new_site["AQI"] = site.get("aqi", "")
        new_site["狀態"] = site.get("status", "")
        filter_data.append(new_site)
    return filter_data

def save_csv(data:list[dict]):
    pprint(data)

def main():
    #下載資料
    try:
         site_list = get_aqi()
    except Exception as e:
        print(e)
        return  #如果在function出錯, 直接return. '提早離開‘
    else:
        print("下載成功!!!")
    
    #過濾資料
    sites = filter_data(site_list)
    
    #儲存資料成csv
    save_csv(sites)
 

if __name__ == "__main__":
    main()

```
