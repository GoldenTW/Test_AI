```python
start_date = "2025-01-01"
end_date = "2025-12-31"
stickers = ["2330.TW", "2303.TW", "2454.TW", "2317.TW"]
data = {ticker: yf.download(ticker, start=start_date, end=end_date, interval="1d",auto_adjust=True) for ticker in stickers}
data
```

```python
import pandas as pd

data1 = {key:dataframe['Close'][key] for key, dataframe in data.items()}
price_df = pd.DataFrame(data1) #講series轉dataframe, key是欄位名稱
price_df1 = price_df.rename(columns={'2330.TW':'台積電', '2303.TW':'聯電', '2454.TW':'聯發科', '2317.TW':'鴻海'})
price_df1
```

```python
price_df1.corr()
```

```python
#要了解每日的漲跌情況, 可以使用pct_change()計算每日的百分比變化
daily_returns = price_df1.pct_change()

#裡面的值乘以100就是百分比
daily_returens_percent = daily_returns * 100
#把所有的nan移除
daily_returens_percent = daily_returens_percent.dropna()
# Date中保留日期,不保留時間
daily_returens_percent.index = daily_returens_percent.index.date

#儲存個內負值顯示綠色, 正值顯示紅色, 0值顯示白色
def color_negative_red(value):
    if value < 0:
        color = 'green'
    elif value > 0:
        color = 'red'
    else:
        color = 'white'
    return f'color: {color}'
daily_returens_percent.style.map(color_negative_red).format("{:.1f}%")

```
