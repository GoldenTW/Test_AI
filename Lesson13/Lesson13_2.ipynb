```python
import matplotlib as plot
from matplotlib.font_manager import fontManager

# 將字型加入 Matplotlib
fontManager.addfont("ChineseFont.ttf")

student1 = {"國文":88, "英文":66, "數學":55, "歷史":75, "探索":85}
student2 = {"國文":92, "英文":70, "數學":80, "歷史":90, "探索":100}

subjects = list(student1.keys())
score1 = list(student1.values())
socre2 = list(student2.values())

# 繪製一個Ｘ為科目, Y為分數的曲線圖
plt.plot(subjects, score1, label="學生A", marker='o')
plt.plot(subjects, socre2, label="學生B", marker='o')
plt.title("學生分數比較", fontname="ChineseFont", fontsize=20)
plt.xlabel("科目", fontname="ChineseFont", fontsize=16)
plt.ylabel("分數", fontname="ChineseFont", fontsize=16)
plt.legend(fontsize=14)
plt.show()

plt.bar(subjects, score1, label="學生A", alpha=0.5)
plt.bar(subjects, socre2, label="學生B", alpha=0.5)
plt.title("學生分數比較", fontname="ChineseFont", fontsize=20)
plt.xlabel("科目", fontname="ChineseFont", fontsize=16)
plt.ylabel("分數", fontname="ChineseFont", fontsize=16)
plt.legend(fontsize=14)
plt.show()  



```

```python
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager

# --- 加入中文字型（假設你已經下載 ChineseFont.ttf） ---
fontManager.addfont("ChineseFont.ttf")
plt.rc('font', family="ChineseFont")
plt.rcParams["axes.unicode_minus"] = False


# --- 學生分數 ---
student1 = {"國文": 88, "英文": 66, "數學": 55, "歷史": 75, "探索": 85}
student2 = {"國文": 92, "英文": 70, "數學": 80, "歷史": 90, "探索": 100}

subjects = list(student1.keys())
score1 = list(student1.values())
score2 = list(student2.values())   # 修正錯字


# --- 畫圖 ---
plt.figure(figsize=(8, 5))

# 第一條線：學生 1
plt.plot(subjects, score1,
         marker='o', linestyle='-', linewidth=2, label="學生 1")

# 第二條線：學生 2
plt.plot(subjects, score2,
         marker='s', linestyle='--', linewidth=2, label="學生 2")

plt.title("學生成績比較")
plt.xlabel("科目")
plt.ylabel("分數")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
