```python
import pandas as pd

df1= pd.read_csv('各鄉鎮市區人口密度.csv')

print(type(df1.columns ))

#轉成list並顯示
col_list= df1.columns.tolist()
print(col_list)
```

```python
df1
```

```python
#更改df的欄位名稱
#使用已存在的cols列表重新命名欄位
df1.columns= df1.loc[0].tolist()
df2=df1.drop(index=0).reset_index(drop=True)
#顯示前10筆資料
df2.head(10)
#修改欄位名稱
df2.columns = ['統計年','鄉鎮市區','人口數','面積(平方公里)','人口密度(人/平方公里)']
df2.head(10)
df2.tail(10) #顯示後10筆資料
df2 = df2[df2.index<=270] #保留前271筆資料
df2 = df2.reindex(index=range(271)) #重設索引值方法,同上效果

df2
```

```python
df2.dtypes
#將'人口數'轉成整數型態
df2['人口數'] = df2['人口數'].astype('int64')
df2['面積(平方公里)'] = df2['面積(平方公里)'].astype('float64')
df2.dtypes
df2
```

```python
#將df2存成新的csv檔案
df2.to_csv('各鄉鎮市區人口密度_整理後.csv', index=False, encoding='utf-8-sig')

```

```python
#將df2存成新的excel檔案
df2.to_excel('各鄉鎮市區人口密度_整理後.xlsx', index=False)
```
