```python
import pandas as pd

series1 = pd.Series([10, 20, 30, 40, 50], index=['a', 'c', 'e', 'd', 'b'])


seriese2 = series1.sort_index()
series1.reindex(sorted(series1.index))
series1.reindex(['a', 'b', 'c', 'd', 'e'])
series1.reindex(['c','d','e'])
```

```python
import numpy as np

frame = pd.DataFrame(
    data=np.arange(9).reshape(3, 3)
    ,index=['a', 'c', 'e']
    ,columns=['Ohio', 'Texas', 'California']    
    )

frame
```

```python
frame.reindex(sorted(frame.index))
frame.reindex(index=['a', 'c'],columns=['California', 'Ohio'])
```

```python
# 實體方法dropna()，可以用來移除缺失值

frame1= pd.DataFrame(
    data=np.arange(16).reshape(4, 4)
    ,index=['台北', '台中', '台南', '高雄']
    ,columns=['one', 'two', 'three', 'four']    
    )
frame1
```

```python
frame1.drop(index=['台北','台中'],columns=['one','two'])
```
