```python
import numpy as np

students=np.random.randint(50,high=101,size=(50,5))
students.shape
```

```python
#ndarray的實體方法

students.sum(axis=1)  #每位學生的總分
students.mean(axis=0)  #每科的平均分

```

```python
# numpy 內建的函式


np.mean(students,axis=0)  #每科的平均分
np.sum (students,axis=1)  #每位學生的總分
```

```python
#1家分店站30天的營收

oneStore = np.random.randint(100,high=100000,size=(30,))
oneStore
```

```python
# 累計每天這家店的營收
import matplotlib.pyplot as plt

oneStoreCum = oneStore.cumsum()
oneStoreCum

# 繪圖累積營收
plt.plot(oneStoreCum)
plt.title("Cumulative Revenue of One Store")
plt.xlabel("day")
plt.ylabel("money")

plt.show()
```

```python
# 5家店30天的營收
fiveStore = np.random.randint(100,high=100000,size=(5,30))
fiveStoreCum = fiveStore.cumsum(axis=1)
# 繪圖累積營收
for i in range(5):
    plt.plot(fiveStoreCum[i]) # 繪製第i家店的累積營收曲線
plt.title("Cumulative Revenue of Five Stores")
plt.xlabel("day")
plt.ylabel("money")    
```
