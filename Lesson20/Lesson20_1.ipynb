```python
import pandas as pd

tips = pd.read_csv('tips.csv')
tips.info()
```

```python
#改變tips的欄位名稱
tips.columns = ['總票價', '小費', '是否吸菸', '日期', '時間', '人數']   
tips
```

```python
#欄位重新排列
tips = tips[['日期', '時間', '是否吸菸', '人數', '總票價', '小費']]

#新鄒一個欄位 小費佔總票價比例
tips['小費佔比'] = tips['小費'] / tips['總票價']
tips
```

```python
# 依照日期,建立一個群組
tips_grouped = tips.groupby('日期')
len(tips_grouped)
```
