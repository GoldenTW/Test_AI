```python
import pandas as pd

tips = pd.read_csv('tips.csv')
tips.info()
```

```python
#改變tips的欄位名稱
tips.columns = ['總票價', '小費', '是否吸菸', '日期', '時間', '人數']   
tips
```

```python
#欄位重新排列
tips = tips[['日期', '時間', '是否吸菸', '人數', '總票價', '小費']]

#新鄒一個欄位 小費佔總票價比例
tips['小費佔比'] = tips['小費'] / tips['總票價']
tips
```

```python
# 依照日期,建立一個群組
tips_grouped = tips.groupby('日期')
len(tips_grouped)
```

```python
#計算小費平均值
tips_grouped['小費'].mean()
#計算小費平均值. 使用dataframe格式呈現
tips_grouped[['小費']].mean()
```

```python
# 依據日期和是否抽煙建立群組
tips_grouped2 = tips.groupby(['日期', '是否吸菸'])

#取得小費的平均值
s1 = tips_grouped2['小費'].mean()

#s1是一個多層索引的Series
#變成DataFrame
df1 = s1.unstack()
df1
print(df1)

# s1我要unstack, 日期在欄位, 是否吸煙在索引
df2 = s1.unstack(level='日期')
print(df2)
print(df2.T)
```

```python
# 依據日期和是否抽煙建立群組
grouped2 = tips.groupby(['日期', '是否吸菸'])
df3 = grouped2['小費'].agg(['mean','sum','count']).unstack(level='日期')
df3.columns.names = ['統計量','日期']

#統計量欄位和日期欄位交換位置, 讓日期在前面, 統計量在後面
df4 = df3.swaplevel(axis=1)
df4.sort_index(axis=1, level=0)  #依照欄位的第一層排序
df4
```
