```python
from tools import all_years_merge
import pandas as pd

all_data = all_years_merge()
all_data

#修改欄位名稱, trnOpDates:乘車日期, gateInComingCnt:進站人數, garteOutGoingCnt:出站人數


# 修改欄位名稱
all_data = all_data.rename(columns={
    'trnOpDate': '乘車日期',
    'gateInComingCnt': '進站人數',
    'gateOutGoingCnt': '出站人數'
})

all_data

# 乘車日期改成時間格式
all_data['乘車日期'] = pd.to_datetime(
    all_data['乘車日期'].astype(str),
    format='%Y%m%d'
)

all_data

all_data['年份'] = all_data['乘車日期'].dt.year
all_data['月份'] = all_data['乘車日期'].dt.month

all_data.head()
```
