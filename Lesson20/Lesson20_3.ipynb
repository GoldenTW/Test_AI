```python
import pandas as pd

df1 = pd.read_csv('tips.csv')
df1.columns = ['總票價','小費','吸煙者','日期','時間','人數']
df1['小費比例'] = df1['小費']/df1['總票價']
df1
```

```python
df1.sort_values(by='小費比例',ascending=False).iloc[:10]
```

```python
def top(df,n=5,column='小費比例'):
    return df1.sort_values(by='小費比例',ascending=False).iloc[:n]

top(df1,n=10)
```

```python
df1.groupby('吸煙者').apply(top,n=5,include_groups=False)
```
