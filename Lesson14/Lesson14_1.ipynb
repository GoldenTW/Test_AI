```python
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager

# 加入自訂的中文字型，避免中文顯示亂碼
fontManager.addfont("ChineseFont.ttf")
# 設定 matplotlib 使用指定字型
plt.rc('font', family="ChineseFont")
# 避免負號顯示為方塊或亂碼
plt.rcParams["axes.unicode_minus"] = False

# 範例資料：兩位學生在各科的分數（字典格式：科目->分數）
student1 = {"國文":88, "英文":66, "數學":55, "歷史":75, "探索":85}
student2 = {"國文":92, "英文":70, "數學":80, "歷史":90, "探索":100}

# 將字典的鍵與值轉為列表，方便繪圖（橫軸為科目，縱軸為分數）
subjects = list(student1.keys())
score1 = list(student1.values())
score2 = list(student2.values())

# 繪製折線圖，並在每個點上加上圓點標記
plt.plot(subjects, score1, label="學生A", marker='o')
plt.plot(subjects, score2, label="學生B", marker='o')

# 設定圖表標題與座標軸標籤，以及字體大小
plt.title("學生分數比較", fontsize=20)
plt.xlabel("科目", fontsize=16)
plt.ylabel("分數", fontsize=16)

# 顯示圖例並加上淡化的格線以便閱讀
plt.legend(fontsize=14)
plt.grid(alpha=0.3)

# 設定 y 軸上下界：以兩組分數的最大/最小值為基準，向外擴展一點空間；並將上界限制為 100
y_min = min(min(score1), min(score2)) - 10
y_max = min(max(max(score1), max(score2)) + 10, 100)
plt.ylim(y_min, y_max)

# 調整標題外觀（距離、底色）
plt.title("學生分數比較", fontsize=20, pad=20, backgroundcolor='yellow')

# 使用 tight_layout 避免標籤被裁切，並設定整體背景色
plt.tight_layout()
plt.gcf().set_facecolor('#e6f2ff')

# 在每個資料點上加上數值標籤，並將文字上移 1 分以免遮蓋點
for i, value in enumerate(score1):
    plt.text(subjects[i], value + 1, str(value), ha='center', fontsize=12)

for i, value in enumerate(score2):
    plt.text(subjects[i], value + 1, str(value), ha='center', fontsize=12)

# 顯示圖表
plt.show()
```
