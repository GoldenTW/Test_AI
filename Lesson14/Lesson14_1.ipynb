```python
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager

# 將字型加入 Matplotlib（你自己的 ChineseFont.ttf）
fontManager.addfont("ChineseFont.ttf")
plt.rc('font', family="ChineseFont")          # 設定中文顯示
plt.rcParams["axes.unicode_minus"] = False   # 避免負號亂碼

student1 = {"國文":88, "英文":66, "數學":55, "歷史":75, "探索":85}
student2 = {"國文":92, "英文":70, "數學":80, "歷史":90, "探索":100}

subjects = list(student1.keys())
score1 = list(student1.values())
score2 = list(student2.values())  # 修正錯字

# 繪製曲線圖
plt.plot(subjects, score1, label="學生A", marker='o')
plt.plot(subjects, score2, label="學生B", marker='o')

plt.title("學生分數比較", fontsize=20)
plt.xlabel("科目", fontsize=16)
plt.ylabel("分數", fontsize=16)

plt.legend(fontsize=14)
plt.grid(alpha=0.3)

# 改變ytick的刻度,由分數的最大值和最小值差距決定. y_max 最大100分. 
y_min = min(min(score1), min(score2)) - 10
y_max = min(max(max(score1), max(score2)) + 10, 100)       
plt.ylim(y_min, y_max)  

#標題距離圖形距離加大,底色改善黃色
plt.title("學生分數比較", fontsize=20, pad=20, backgroundcolor='yellow')  



plt.show()
```
