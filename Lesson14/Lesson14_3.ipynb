```python
import numpy as np
import matplotlib.pyplot as plt

#指定間隔數量,傳出固定間隔的數字
x=np.linspace(-3, 3, 50)

y1=2*x + 1
y2=x**2

plt.plot(x,y1,color='red',label='y=2x+1',linewidth=2,linestyle='--')
plt.plot(x,y2,color='blue',label='y=x^2',linewidth=2,linestyle='-')

plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend()
plt.xlim((-1,2))
plt.ylim((-2,3))
plt.grid()
plt.title('Two Functions Plot')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t=np.arange(0.,5.,0.2)
y1=np.sin(2*np.pi*t)
y2=np.exp(-t)

plt.subplot(2,1,1)
plt.plot(t,y1,'ro-',label='sin(2πt)')
plt.title('Sine Function')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid()
plt.legend()
plt.subplot(2,1,2)
plt.plot(t,y2,'bs--',label='exp(-t)')
plt.title('Exponential Decay')
plt.xlabel('Time (s)')
plt.ylabel('Value')
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```
