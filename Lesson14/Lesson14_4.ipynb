```python
import matplotlib as plot
from matplotlib.font_manager import fontManager

# 將字型加入 Matplotlib，請將自己的字型檔放在同一目錄或提供完整路徑
fontManager.addfont("ChineseFont.ttf")
plot.rc('font',family="ChineseFont")
```

```python
import matplotlib.pyplot as plt

index = ['apple','banna','orange','tomato','guave']
value = [5,7,3,4,6]

figure = plt.figure()
#製作條狀圖
plt.bar(index, value, color='skyblue')
#設定標題與字型
plt.title('水果銷售量', fontname='ChineseFont', fontsize=20)
#設定X軸標籤與字型
plt.xlabel('水果種類', fontname='ChineseFont', fontsize=14)
#設定Y軸標籤與字型
plt.ylabel('銷售量', fontname='ChineseFont', fontsize=14)
#顯示圖表
plt.show() 
```

```python
import numpy as np

index = np.array([3,4,5,6,7])
subject = ['apple','banana','oriange','tomato','guava']
value1 = [5000,6000,7000,8000,9000]
value2 = [3000,4000,5000,6000,7000]
value3 = [4000,5000,6000,7000,8000]

figure = plt.figure()
axes = figure.add_subplot()
bin_width = 0.3
axes.bar(index,value1,width=bin_width,color='red')
axes.bar(index+bin_width,value2,width=bin_width,color='blue')
axes.bar(index-bin_width,value3,width=bin_width,color='green')

axes.set_xticks(index,labels=subject)

#取得x,y軸最小值和最大值

xmin,xmax,ymin,ymax = axes.axis()

xmin -= 0.2
xmax += 0.2
ymin = 1000
ymax += 1000

axes.axis(xmin=xmin,xmax=xmax,ymin=ymin,ymax=ymax)

#顯示圖例legend

axes.legend(labels=['2018年','2019年','2020年'])

#每個bin上顯示數值
for i in range(len(index)):
    axes.text(index[i],value1[i]+100,str(value1[i]),ha='center',fontsize=7)
    axes.text(index[i]+bin_width,value2[i]+100,str(value2[i]),ha='center',fontsize=7)
    axes.text(index[i]-bin_width,value3[i]+100,str(value3[i]),ha='center',fontsize=7)


plt.show()
```
